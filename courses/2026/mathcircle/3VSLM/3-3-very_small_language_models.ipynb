[Project Gutenberg](https://www.gutenberg.org/) (named after the [Gutenberg printing press](https://en.wikipedia.org/wiki/Printing_press)) is a library of old books, whose copyright has expired, meaning that they are free to access.
The following code reads the text of [Alice's Adventures in Wonderland, by Lewis Carroll](https://www.gutenberg.org/cache/epub/11/pg11.txt), does some formatting, and prints the first few lines.

(Fun fact: [Lewis Carroll](https://en.wikipedia.org/wiki/Lewis_Carroll#Mathematical_work) was also a mathematician! He worked in geometry, linear algebra, and mathematical logic.)

```python
import re
import urllib.request
import random

URL = "https://www.gutenberg.org/cache/epub/11/pg11.txt"
raw = urllib.request.urlopen(URL).read().decode("utf-8-sig")

# remove the Gutenberg header and licence using the marker lines
start = re.search(r"\*\*\* START OF .*? \*\*\*", raw).end()
end = re.search(r"\*\*\* END OF .*? \*\*\*", raw).start()
text = raw[start:end]

# make text lowercase (so we treat "the" and "The" as the same word)
text = text.lower()

# remove punctuation
text = text.replace("’", "'").replace("‘", "'")   # make apostrophes consistent
text = re.sub(r"[—–-]+", " ", text)               # replace dashes with spaces to avoid gluing words
text = re.sub(r"[^a-z0-9'\s]", "", text)          # drop all other punctuation except apostrophes

# split into words and strip stray quote marks from word edges
words = [w.strip("'") for w in text.split()]
words = [w for w in words if w]

print(f"there are {len(words)} words and we've stored them in a list that looks like:")
print(words[94:150])
```
