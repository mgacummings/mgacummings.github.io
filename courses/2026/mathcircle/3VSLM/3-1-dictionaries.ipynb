## Part 1. Python dictionaries

We start with a new data type in python, called a dictionary.
A dictionary can be thought of as a look-up table: given an input, it looks up an output.

Here's an example, where our dictionary looks up prices of items in a store.

```python
prices = {
    "Chefman Rotating Waffle Maker": 60,
    "Ninja Creami Ice Cream Maker": 230,
    "Ooni Koda 12 Pizza Oven": 550
}

print(prices)
```

We can look up the price of an item as follows:

```python
prices["Ooni Koda 12 Pizza Oven"]
```

We can also add in other entries later:

```python
prices["KitchenAid Stand Mixer"] = 400
```

```python
print(prices)
```

```python
prices["KitchenAid Stand Mixer"]
```

Here's another way we can use dictionaries.
Imagine that we want to keep track of a list of contacts: For each contact, we have a phone number, an email address, and a street address.

We'll store this in a dictionary where for each name, we'll have a list containing the phone number, email address, and street address.

```python
contacts = {
    "Alice": ["905-845-5200", "alice@wonderland.com", "1330 Montclair Dr"],
    "Bob": ["416-341-1000", "bob@builder.org", "1 Blue Jays Way"],
    "Charlie": ["613-993-9620", "charlie@chaplin.com", "316 Mont-Royal Ave"]
}
```

```python
print(contacts)
```

```python
contacts["Bob"]
```

📌 Add a new contact and look up their information

Imagine that we have a really large dictionary, so big that we can't look through all the entries one-by-one.
But we can ask whether a given entry exists:

```python
"Olivia" in contacts
```

Lastly, imagine that we now want to keep track of another piece of information for each contact, such as a birthday.
Since we're using lists, we can still use the `.append()` method.

```python
contacts["Charlie"].append("16 April 1889")
```

```python
print(contacts)
```

```python
print(contacts["Charlie"])
```
