## Part 2. Predator-prey population modelling

Let $P_N$ and $S_N$ be the population of penguins and seals in year $N$

$$\textsf{next year's populations:} \quad
\begin{cases}
  P_{N+1} = r_p P_N + uS_N \\
  S_{N+1} = vP_N + r_s S_N
\end{cases}
$$


* $r_p$ and $r_s$ are the **reproduction rates** for penguins and seals

* $u$ and $v$ are cross-species interaction parameters

P2 = (penguin reproduction)P1 + uS1
= 1.1(93) - 0.2(33)
= 102.3 - 6.6
= 95.7

S2 = vP1 + (seal reproduction)S1
= 0.1(93) + 0.8(33)
= 9.3 + 26.4
= 35.7

```python
# run this cell, but you don't need to read it (so skip to the next cell)
# we'll use this function to graph our computations

import matplotlib.pyplot as plt

def plot_populations(penguins, seals):
    # plots popluations of penguins and seals on the same graph

    plt.figure(figsize=(10, 6))

    time_steps = list(range(len(penguins)))
    plt.plot(time_steps, penguins, marker='o', linestyle='-', label="penguins")
    plt.plot(time_steps, seals, marker='^', linestyle='-', label="seals")

    plt.xlabel('Years')
    plt.ylabel('Population')
    plt.title('Penguin and Seal Popluations')
    plt.grid(True)
    plt.legend()
    plt.show()
```

📌 Fill in the following function. Its inputs are
* initial populations `P0` and `S0` for penguins and seals, respectively

* reproduction rates `rp` and `rs` for penguins and seals, respectively

* cross-species interaction terms `u` and `v`, defined as above

* the number of years to simulate `years`

The function should output two lists whose entries are the yearly population values.

```python
def model_populations(P0, S0, rp, rs, u, v, years):

  # start here, by defining lists
  penguins = [P0]
  seals = [S0]

  for i in range(years):
    next_penguin_pop = rp*penguins[-1] + u*seals[-1]
    next_seal_pop = v*penguins[-1] + rs*seals[-1]

    penguins.append(next_penguin_pop)
    seals.append(next_seal_pop)

  return (penguins, seals)

# run this cell once your function is ready!
```

```python
# run this cell, if everything's working, you should see a graph

initial_penguins = 90
initial_seals = 30

penguin_reproduction = 1.1
seal_reproduction = 0.8

u = -0.3
v = 0.2

years = 20

(penguin_pop, seal_pop) = model_populations(initial_penguins, initial_seals, penguin_reproduction, seal_reproduction, u, v, years)
plot_populations(penguin_pop, seal_pop)
```
