## Part 1. Single-species population modelling

Let $P_N$ be the penguin population in year $N$
$$\textsf{next year's population: } P_{N+1} = rP_N + m$$


* $r$ is the **reproduction rate**

* $m$ is the **migration rate**

```python
# run this cell, but you don't need to read it (so skip to the next cell)
# we'll use this function to graph our computations

import matplotlib.pyplot as plt

def plot_population(*populations):
    # plots populations on one graph

    plt.figure(figsize=(10, 6))

    for i, P in enumerate(populations):
        time_steps = list(range(len(P)))
        lbl = f'Population {i+1}'
        plt.plot(time_steps, P, marker='o', linestyle='-', label=lbl)

    plt.xlabel('Years')
    plt.ylabel('Population')
    plt.title('Penguin Popluation')
    plt.grid(True)
    plt.legend()
    plt.show()
```

📌 Fill in the following function. Its inputs are
* initial population `P0`

* reproduction rate `r`

* migration rate `m`

* the number of years to simulate `years`

The function should output a list whose entries are the yearly population values.
You'll want to use some sort of loop.

You might want to look at some of last week's files to remind yourself of the python formatting and grammar

```python
def model_population(P0, r, m, years):
  population = [P0]  # start by defining a list containing the starting population value P0; we'll add the other population values to it later

  for i in range(years):
    # P_(N+1) = r*P_N + m
    # most recent population value is the last entry in population list
    # last entry in the population list is population[-1]
    next_population_value = r * population[-1] + m
    population.append(next_population_value)

  return population

# don't forget to run this cell once you've written your function!
```

```python
# run this cell, if everything's working, you should see a graph

initial_pop = 100
migration_rate = 5
years = 15

reproduction_rate = 1.0194
pop_data = model_population(initial_pop, reproduction_rate, migration_rate, years)

plot_population(pop_data)
```

```python
# here are some different reproduction rates
# run this cell, and see some different behaviours

reproduction_rate1 = 1.0194
pop_data1 = model_population(initial_pop, reproduction_rate1, migration_rate, years)

reproduction_rate2 = 0.9676
pop_data2 = model_population(initial_pop, reproduction_rate2, migration_rate, years)

reproduction_rate3 = 0.8618
pop_data3 = model_population(initial_pop, reproduction_rate3, migration_rate, years)

plot_population(pop_data1, pop_data2, pop_data3)
```
