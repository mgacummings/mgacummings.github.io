## Part 3. SIR model in epidemiology

Let $S_N$, $I_N$, and $R_N$ be the populations of susceptible, infectious, and recovered individuals in month $N$

$$\textsf{next month's populations:} \quad
\begin{cases}
  S_{N+1} \hspace{0.3mm} = S_N - aS_NI_N \\
  I_{N+1} \hspace{1.4mm} = I_N + aS_NI_N - bI_N \\
  R_{N+1} = R_N + bI_N
\end{cases}
$$

* $a$ and $b$ are **transmission rates** (these will be fractions between 0 and 1)

```python
# run this cell, but you don't need to read it (so skip to the next cell)
# we'll use this function to graph our computations

import matplotlib.pyplot as plt

def plot_SIR(susceptible, infectious, recovered):
    # plots the susceptible, infectious, and recovered populations over time

    plt.figure(figsize=(10, 6))

    time_steps = list(range(len(susceptible)))
    plt.plot(time_steps, susceptible, marker='o', linestyle='-', label="susceptible")
    plt.plot(time_steps, infectious, marker='^', linestyle='-', label="infectious")
    plt.plot(time_steps, recovered, marker='+', linestyle='-', label="recovered")

    plt.xlabel('Months')
    plt.ylabel('Population')
    plt.title('Infectious Disease Modelling')
    plt.grid(True)
    plt.legend()
    plt.show()
```

* SIS (susceptible-infectious-susceptible)
