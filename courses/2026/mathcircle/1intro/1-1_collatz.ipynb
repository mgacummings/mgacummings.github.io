### 1. Collatz conjecture

Recall the following procedure for the Collatz conjecture:

Let $n$ be a positive integer.
  * if $n = 1$, then stop
  * if $n$ is even, then return $n/2$
  * otherwise, return $3n+1$

---

📌 Fill in the following function `collatz_step`: it takes in a positive integer `n`, and returns the output as prescribed above.

Notes:
  * the "base case" (when $n = 1$) is already filled in
  * to check whether a number is even, think about how to use the `%` operator, which returns "remainder after division", for example, `7 % 3` evaluates to `1`, because $7 = 2\times 3 + 1$, so the remainder is 1

```python
def collatz_step(n):

  # the double equal sign "==" checks whether the left and right sides are equal
  if n == 1:
    return n

  if n % 2 == 0:  # is n even?
    return n//2

  if n % 2 == 1:  # is n odd?
    return 3*n+1


  # when you're done, press shift+enter to "run" this cell, so you can call this function from the next cell
```

```python
# test your code by running (shift+enter) this cell

print(collatz_step(1))  # should print 1
print(collatz_step(5))  # should print 16
print(collatz_step(16))  # should print 8
```

Let's automate this!
We can use a `while` loop to repeat something until we're done.

Here's an example.
The _factorial_ of a number is defined by
$$n! = n\times (n-1) \times (n-2) \times \cdots \times 2 \times 1,$$
or equivalently,
$$n! = n\times (n-1)! \quad\text{where}\quad 1! = 1$$

The following code finds the smallest number whose factorial is more than 1000.

2! = 2 x 1 = 2

3! = 3 x 2 x 1 = 6

4! = 4 x 3 x 2 x 1 = 4 x (3!) = 4 x 6 = 24

```python
n = 1
factorial = 1  # 1! = 1

while factorial < 1000:
  n = n+1  # this replaces n by n+1, but it's still named n (so if n was 2, we've updated n to be 3)
  factorial = n * factorial  # this is the second definition of factorial

print(n)  # this prints the answer, after the loop has finished
```
