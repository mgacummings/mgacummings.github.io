### 2. Approximating $\pi$

Recall that if a circle is inscribed within a square, and a random point is picked within the square, then the ratio $r$ that this point is also within the circle is $$r = \pi/4.$$

So if we can approximate $r$ by running lots of experiments, then we can solve for $\pi = 4r$.

---

Our circle has radius $1$, so if it is centred at the origin $(0, 0)$, then our square contains points $(x, y)$ where $x$ and $y$ are both real numbers between $-1$ and $1$.
(see [this sketch on desmos](https://www.desmos.com/calculator/erlw2pfg7i))

📌 Write a function `random_point` that returns a random point `(x, y)` in the square.

```python
import random  # run this cell
```

```python
# this produces a random real number between 0 and 8
x = random.uniform(0, 8)  # "uniform" means all values from 0 to 8 are equally likely
print(x)
```

```python
def random_point():

  # fill in this code
  x = random.uniform(-1, 1)
  y = random.uniform(-1, 1)

  return (x, y)
```

```python
# check your code works (if you re-run this cell, you should get a different value)
print(random_point())
```

Now we want a function that tests whether a given point lies within the circle.

Our circle contains all points that are within _distance 1_ of the origin, and we can use the Pythagorean theorem to compute this distance!
The distance of the point $(x, y)$ to the origin is $$\sqrt{x^2 + y^2},$$
and to see why, see [this illustration on desmos](https://www.desmos.com/calculator/wjqj5odyjy).

📌 Write a function `is_in_circle(x, y)` to determine whether the given point `(x, y)` lies within distance 1 of the origin, using the formula above.

You'll need the `math` function `sqrt` to take square roots, so we'll import that here:

```python
from math import sqrt
```

```python
# try it out:
sqrt(9)
```

Also, you might use an `if` statement like we had in the first notebook.
You can use `a <= b` to test whether $a \le b$.
Your code might `return True` or `return False` depending on the distance you compute.

```python
def is_in_circle(x, y):

  distance = sqrt(x*x + y*y)

  if distance <= 1:
    return True
  if distance > 1:
    return False
```

We'll now put these two functions together to compute our approximation of the probability $p$.
To get a good approximation, we want to sample lots of points.

We'll use a `for` loop; it's similar to a `while` loop, but it runs a prescribed number of times.
For instance, the following `for` loop runs exactly five times:

```python
for i in range(5):
  print(i)
```

📌 Initialize a variable `circle_hits`, which will count the number of random points that have "hit" (or lie in) our circle.
Then, use a `for` loop as follows: each time through the `for` loop, sample a point, and increment `circle_hits` if the point is in the circle.

```python
num_points = 1000000
circle_hits = 0

for i in range(num_points):
  (x, y) = random_point()
  if is_in_circle(x, y):
    circle_hits = circle_hits + 1
```

📌 Now compute the probability as a percentage of the points that lie in the circle (`circle_hits`), compared to the total number of points sampled (`num_points`). Then, use this to approximate $\pi$.

```python
ratio =  circle_hits / num_points
```

```python
pi = 4*ratio # what was the relation between pi and the `ratio`?
print(pi)
```
