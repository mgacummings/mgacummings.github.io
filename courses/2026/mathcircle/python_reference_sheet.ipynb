## 2. Functions

The line $y = 3x + 4$ can be thought of as a function: it takes in a number $x$, and outputs the number $3x+4$.
So for example, when $x=2$, it outputs $10$.

We can implement this idea into a python function.

📌 Run this cell and the following cell.

````python
def equation_of_line(x):
  y = 3*x + 4
  return y
````

````python
print(equation_of_line(2))
````



*   The first line `def equation_of_line(x):` defines (`def`) a function called `equation_of_line`, which takes one input `x`
*   `y = 3*x + 4` defines a new variable called `y`, and assigns it the value `3*x+4` (the asterisk `*` means multiplication)
*   `return y` means that the function _returns_ the value `y` we just defined, so we can save it, or print it (as above)

Also note that the colon `:` at the end of the first line matters, as does the indentation of lines two and three.



Here's another example.
A number is **even** if it is divisble by 2, or equivalently, if after dividing by 2, there is no remainder.

For example, if we divide $9$ by $2$, we get
$$9 = 2\cdot 4 + 1,$$
which has remainder 1, so $9$ is not even.

Python computes the remainder upon division using the `%` operator.
We can use this to write a function to test whether a number is even.



````python
def is_even(n):
  if n % 2 == 0:  # the double equal sign == asks whether the left-hand side is equal to the right-hand side
    return True
  else:
    return False
````

````python
print(is_even(9))
````

````python
print(is_even(1234))
````

## 3. If this, then that

The previous example uses an `if` statement to return different values depending on the remainder upon division by 2.

The general formatting is as follows:

```
if condition:
  ## do something
else:
  ## do something different
```

Here's a function that returns `"big"` if the input is big (at least 10), and `"small"` otherwise.

📌 Run the next three cells

````python
def size_of_number(n):
  if n >= 10:
    return "big"
  else:
    return "small"
````

````python
print(size_of_number(10))
````

````python
print(size_of_number(-11))
````

Maybe you want to consider more than two cases, in which case you can use this formatting:

```
if condition1:
  ## do something
elif condition2:
  ## do something else
elif condition3:
  ## do another something else
else:
  ## do yet another thing
```
You can have as many `elif` ("else if") statements as you like.

Here's another example, which returns...
* "very big" if the number is at least 100,
* "big" if it's at least 10 but less than 100,
* "small" if it's less than 9 and bigger than 0,
* "zero" if it's 0, or,
* "negative" if it's less than 0

📌 Run the following cells

````python
def size_of_number(n):
  if n >= 100:
    return "very big"
  elif 10 <= n < 100:
    return "big"
  elif 0 < n < 10:
    return "small"
  elif n == 0:
    return "zero"
  else:
    return "negative"
````

````python
print("the first number is 101, which is", size_of_number(101))
print("the second number is 0, which is", size_of_number(0))
print("the third number is 0.5, which is", size_of_number(0.5))
print("the fourth number is -11, which is", size_of_number(-11))
print("the last number is 50, which is", size_of_number(50))
````

## 4. Loops

If we want to call a function with the inputs 1, 2, 3, 4, ..., instead of doing each call manually, we can use a _loop_.

There are a couple different types of loop.
This first one is called a _for loop_, because it runs once _for_ each number in a specified range.

````python
for i in range(1, 5):
  print(i)
````

Notice that in this loop we use `range(1, 5)`, which gives `i=1`, then `i=2`, `i=3`, and `i=4`. (5 is not included!)

Here's another example:

````python
for j in range(-1, 4):
  x = 5*j
  print("the number is", x, "which is", size_of_number(x))
````

If we're not sure how many times we want to loop, we can use a `while` loop.

Here's an example, to find the first "big" number.

````python
i = 0

while size_of_number(i) != "big":
  i = i+1  ## this redefines i to take the value i+1 (so if i was 1, then i becomes 2)

# after the loop, we can find out what the first big number is:
print("the first big number is", i)
````

Here's another example, with the equation of the line $y = 3x+4$ from earlier.

This loop finds the first $x$ such that $y = 3x+4$ is at least 50.

````python
x = 0
y = equation_of_line(x)

while y < 50:
  # print the current values of x and y
  print("the output for x =", x, "is y =", y)

  # update x and y
  x = x+1
  y = equation_of_line(x)

# when the loop ends, we have our answer:
print("when x =", x, "we get y = ", y, "which is at least 50")
````

While loops can behave like for loops.

Recall this for loop from earlier:
```python
for j in range(-1, 4):
  x = 5*j
  print("the number is", x, "which is", size_of_number(x))
```
The next cell gives an equivalent version with a while loop.

````python
j = -1

while j < 4:
  x = 5*j
  print("the number is", x, "which is", size_of_number(x))

  j = j+1
````

## 5. Lists

To store lots of data, we can put it in a list using square brackets `[...]`.

````python
my_first_list = [5, 1, 2, 4, 3]

print(my_first_list)
````

We can aggregate some information from lists:

````python
print("my list is", my_first_list)                        # prints the list again
print("it has length", len(my_first_list))                # prints the length of the list, using len()
print("its maximum value is", max(my_first_list))         # prints the biggest elements, using max()
print("the sum of its elements is", sum(my_first_list))   # prints the sum of its elements, using sum()
````

We can add an element to a list using `.append()`

````python
second_list = [9, 0, 2, 1]
print("at the beginning, the list is", second_list)

second_list.append(0)
print("after appending, our list is", second_list)
````

Notice what happens if we run the following cell multiple times:

````python
second_list.append(50)
print(second_list)
````

To get the first entry from a list `L`, we write `L[0]`.
To get the second entry, we use `L[1]`, the third is `L[2]`, and so on... _(this is called zero-indexing)_

````python
print("the first entry of the list", my_first_list, "is", my_first_list[0])
print("the fourth entry of the list", my_first_list, "is", my_first_list[3])
````

Sometimes we'll want to work backwards, and get the last entries of a list.
We can do this with a negative index: The last entry of the list `L` is `L[-1]`, the second last entry is `L[-2]`, and so on...

````python
print("the last entry of the list", my_first_list, "is", my_first_list[-1])
print("the second-last entry of the list", my_first_list, "is", my_first_list[-2])
````

## Sequences

Sequences are similar to lists in that they store multiple values.
They use round parentheses `(...)` instead of square brackets `[...]`.

````python
my_first_sequence = (5, 1, 2, 4, 3)
print(my_first_sequence)
````

Many of the functions for lists work the same for sequences.

````python
print("my sequence is", my_first_sequence)                    # prints the sequence
print("it has length", len(my_first_sequence))                # prints the length of the list, using len()
print("its maximum value is", max(my_first_sequence))         # prints the biggest elements, using max()
print("the sum of its elements is", sum(my_first_sequence))   # prints the sum of its elements, using sum()
````

We can also extract elements in the same as with lists.

````python
print("the second entry of the sequence", my_first_sequence, "is", my_first_sequence[1])
print("the second-last entry of the sequence", my_first_sequence, "is", my_first_sequence[-2])
````

The main difference is that we can't append to a sequence---once we define it, we can't modify it (instead, we would have to redefine a new sequence).

So the following code throws an error:

````python
my_first_sequence.append(6)
````
